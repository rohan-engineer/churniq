# ChurnIQ: Exploratory Data Analysis
## Telecom Customer Churn Dataset Investigation
This notebook demonstrates exploratory data analysis (EDA) on the IBM Telco Customer Churn dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print('Dataset Shape:', df.shape)
df.head()
```

### Data Cleaning and Type Coercion
`TotalCharges` has empty spaces `' '` for customers with `tenure == 0`.

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)
print('Missing values after cleaning:\n', df[['tenure', 'MonthlyCharges', 'TotalCharges']].isna().sum())
```

### Overall Churn Ratio

```python
churn_counts = df['Churn'].value_counts()
print(churn_counts)
plt.figure(figsize=(6, 4))
churn_counts.plot(kind='pie', autopct='%1.1f%%', colors=['#2563eb', '#ef4444'], explode=[0, 0.08])
plt.title('Overall Customer Churn')
plt.ylabel('')
plt.show()
```

### Churn by Contract Type

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Contract', hue='Churn', palette=['#2563eb', '#ef4444'])
plt.title('Churn by Contract Type')
plt.show()
```

### Churn by Payment Method

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, y='PaymentMethod', hue='Churn', palette=['#2563eb', '#ef4444'])
plt.title('Churn by Payment Method')
plt.show()
```

### Tenure and Monthly Charges

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, ax=ax[0], palette=['#2563eb', '#ef4444'])
ax[0].set_title('Tenure Distribution')
sns.boxplot(data=df, x='Churn', y='MonthlyCharges', ax=ax[1], palette=['#2563eb', '#ef4444'])
ax[1].set_title('Monthly Charges vs Churn')
plt.tight_layout()
plt.show()
```
